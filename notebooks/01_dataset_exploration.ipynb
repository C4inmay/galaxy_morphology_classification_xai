```python
galaxy_labels = {
    "Class1.1": "Smooth",
    "Class1.2": "Features or Disk",
    "Class1.3": "Star / Artifact",

    "Class2.1": "Edge-on",
    "Class2.2": "Not Edge-on",

    "Class3.1": "Bar",
    "Class3.2": "No Bar",

    "Class4.1": "Spiral",
    "Class4.2": "No Spiral",

    "Class5.1": "No Bulge",
    "Class5.2": "Just Noticeable",
    "Class5.3": "Obvious",
    "Class5.4": "Dominant",

    "Class6.1": "Odd",
    "Class6.2": "Not Odd",

    "Class7.1": "Completely Round",
    "Class7.2": "In Between",
    "Class7.3": "Cigar-shaped",

    "Class8.1": "Ring",
    "Class8.2": "Lens/Arc",
    "Class8.3": "Disturbed",
    "Class8.4": "Irregular",
    "Class8.5": "Other",
    "Class8.6": "Merger",
    "Class8.7": "Dust Lane",

    "Class9.1": "Tight Arms",
    "Class9.2": "Medium Arms",
    "Class9.3": "Loose Arms",

    "Class10.1": "1 Arm",
    "Class10.2": "2 Arms",
    "Class10.3": "3+ Arms",

    "Class11.1": "Round Bulge",
    "Class11.2": "Boxy Bulge",
    "Class11.3": "No Bulge",
    "Class11.4": "Irregular Bulge",
    "Class11.5": "Other Bulge",
    "Class11.6": "Unknown Bulge"
}
```

```python
for key, value in galaxy_labels.items():
    print(f"{key:10} --> {value}")
```

```python
question_groups = {
    "Q1": ["Class1.1", "Class1.2", "Class1.3"],
    "Q2": ["Class2.1", "Class2.2"],
    "Q3": ["Class3.1", "Class3.2"],
    "Q4": ["Class4.1", "Class4.2"],
    "Q5": ["Class5.1", "Class5.2", "Class5.3", "Class5.4"],
    "Q6": ["Class6.1", "Class6.2"],
    "Q7": ["Class7.1", "Class7.2", "Class7.3"],
    "Q8": ["Class8.1", "Class8.2", "Class8.3", "Class8.4", "Class8.5", "Class8.6", "Class8.7"],
    "Q9": ["Class9.1", "Class9.2", "Class9.3"],
    "Q10": ["Class10.1", "Class10.2", "Class10.3"],
    "Q11": ["Class11.1", "Class11.2", "Class11.3", "Class11.4", "Class11.5", "Class11.6"],
}
```

```python
for question, columns in question_groups.items():
    df[question] = df[columns].idxmax(axis=1)

df[["Q1", "Q2", "Q3", "Q4", "Q5"]].head()
```

```python
for question in question_groups.keys():
    df[question] = df[question].map(galaxy_labels)

df[["Q1", "Q2", "Q3", "Q4", "Q5"]].head()
```

```python
df[["Q1","Q2","Q3","Q4","Q5","Q6","Q7","Q8"]].head()
```

```python
def assign_class(row):

    # Ignore stars/artifacts
    if row["Q1"] == "Star / Artifact":
        return "Star"

    # Smooth galaxies
    if row["Q1"] == "Smooth":
        if row["Q7"] == "Completely Round":
            return "Elliptical"

        elif row["Q7"] == "In Between":
            return "Oval"

        elif row["Q7"] == "Cigar-shaped":
            return "Cigar"

    # Disk galaxies
    if row["Q1"] == "Features or Disk":

        if row["Q2"] == "Edge-on":
            return "Edge-on Spiral"

        if row["Q4"] == "Spiral":

            if row["Q3"] == "Bar":
                return "Barred Spiral"

            return "Spiral"

        if row["Q8"] == "Ring":
            return "Ring Galaxy"

        if row["Q8"] == "Merger":
            return "Merger"

        if row["Q8"] == "Irregular":
            return "Irregular"

        if row["Q8"] == "Dust Lane":
            return "Dust Lane"

        return "Disk"

    return "Other"
```

```python
df["Galaxy_Class"] = df.apply(assign_class, axis=1)

df[["GalaxyID", "Galaxy_Class"]].head()
```

```python
class_counts = df["Galaxy_Class"].value_counts()

class_counts
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

class_counts.plot(kind="bar")

plt.title("Galaxy Class Distribution")
plt.xlabel("Galaxy Class")
plt.ylabel("Number of Images")

plt.xticks(rotation=45)

plt.show()
```

```python
df["Galaxy_Class"] = df["Galaxy_Class"].replace({
    "Star": "Other",
    "Dust Lane": "Other"
})
```

```python
class_counts = df["Galaxy_Class"].value_counts()

class_counts
```
